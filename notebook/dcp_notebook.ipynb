## Dropping **"Peak"** and **"All Time Peak"**

```python
df.drop(columns=["Peak", "All Time Peak"],inplace = True)
```

## Fixing Inconsistent Formats

### Standralizing Test Casing for All String Columns

```python
cols = ["Artist", "Tour title"]

df[cols] = df[cols].apply(lambda col: col.str.strip().str.title())
```

```python
for col in df.columns:
    print(repr(col))
```

### Fixing DataTypes

```python
c = ["Actual gross", "Adjusted gross (in 2022 dollars)", "Average gross","Ref."]

df[c] = df[c].apply(
    lambda col: pd.to_numeric(
        col.str.replace(r"[^\d]", "", regex=True),
        errors="coerce"
    )
)
```
